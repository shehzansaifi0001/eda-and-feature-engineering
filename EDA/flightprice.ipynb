```python
df['date']=df['Date_of_Journey'].str.split('/').str[0].astype(int)
df['month']=df['Date_of_Journey'].str.split('/').str[1].astype(int)
df['year']=df['Date_of_Journey'].str.split('/').str[1].astype(int)
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Departure_hour']=df['Dep_Time'].str.split(':').str[0].astype(int)
df['Departure_min']=df['Dep_Time'].str.split(':').str[1].astype(int)
```

```python
df.info()
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0, '2 stops':2, '1 stop':1, '3 stops':3, np.nan:1, '4 stops':4})
```

```python
df['Total_Stops']
```

```python
df.drop('Route',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Duration']=df['Duration'].str.split(' ').str[0].str.split('h').str[0]
```
